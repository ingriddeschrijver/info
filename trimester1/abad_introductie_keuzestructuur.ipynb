```python
woord = input("Geef een woord: ")     
aantal_letters = len(woord)
if aantal_letters == 6: 
    tussenvoegsel = 'exact'
elif aantal_letters < 6:    
     tussenvoegsel = 'minder dan'
else:
     tussenvoegsel = 'meer dan'
print(f'Het woord \'{woord}\' bevat {tussenvoegsel} 6 karakters.')
```

```python
rico1 = float(input('Geef de rico van de eerste rechte: '))
rico2 = float(input('Geef de rico van de tweede rechte: '))
if rico1 == rico2:
    stand = 'evenwijdig'
else:
    stand = 'snijdend'
print(f'De rechten zijn {stand}.')
      
```

```python
rico = float(input('Geef de rico van de rechte: '))

if rico > 0:
    verloop = 'stijgend'
elif rico < 0:
    verloop = 'dalend'
else:
    verloop = 'een horizontale'
print(f'De rechte is {verloop}.')
```
