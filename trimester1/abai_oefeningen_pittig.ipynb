- Let bij f-strings goed op spaties, haakjes en leestekens.

### Oefening 18

Voorspel de uitvoer. Verwacht je een foutmelding? Schrijf dan gewoon "fout".

````python
saldo = 50
for bedrag in [-20, 35, -70, 10]:
    if saldo + bedrag >= 0:
        saldo += bedrag
        print(f"ok: {bedrag} -> {saldo}")
    else:
        print(f"geweigerd: {bedrag}")
print(f"Eindsaldo: {saldo}")
````

### Oefening 25

Voorspel de uitvoer. Verwacht je een foutmelding? Schrijf dan gewoon "fout".

````python
for i in range(1, 4):
    lijn = ""
    for j in range(4 - i):
        lijn += "*"
    print(f"{i}{lijn}")
````
