### Oefening 1

Voorspel de uitvoer. Verwacht je een foutmelding? Schrijf dan gewoon "fout".

````python
def verdrievoudigen(x):
    """Verdrievoudigen."""
    return 3 * x

print(verdrievoudigen(4), verdrievoudigen("ab"), verdrievoudigen(1.5))
````

**Uitvoer:**
```

```

**Uitleg:**

De functie geeft telkens `3 * x` terug. Wat `*` doet, hangt af van het **type** van het argument:

| oproep | `x` | type | `3 * x` | resultaat |
|---|---|---|---|---|
| `verdrievoudigen(4)` | 4 | int | 3 · 4 | `12` |
| `verdrievoudigen("ab")` | "ab" | str | tekst 3 keer **herhalen** | `ababab` |
| `verdrievoudigen(1.5)` | 1.5 | float | 3 · 1,5 | `4.5` |

- Eén en dezelfde functie werkt dus voor verschillende types.
- `print` met komma's zet een spatie tussen de drie waarden.

### Oefening 2

````python
def kwadraat(x):
    """Kwadraat van x."""
    print(x ** 2)

resultaat = kwadraat(5)
print(resultaat)
````


**Uitleg:**

| stap | wat gebeurt er? | uitvoer |
|---|---|---|
| 1 | `kwadraat(5)` wordt opgeroepen: `x = 5` | – |
| 2 | **in** de functie: `print(x ** 2)` | `25` |

- ⚠️ `print` en `return` zijn niet hetzelfde:
  - `print` **toont** iets op het scherm, maar het resultaat is daarna "weg".
  - `return` **geeft** een waarde **terug** aan wie de functie oproept, zodat je ermee verder kan rekenen of ze kan opslaan.
- Een functie zonder `return` geeft automatisch `None` ("niets") terug.

### Oefening 4

Voorspel de uitvoer. Verwacht je een foutmelding? Schrijf dan gewoon "fout".

````python
def volgende(x):
    """Het volgende getal."""
    return x + 1
    print("klaar")

def volgende_bis(x):
    """Het volgende getal."""
    print("klaar")
    return x + 1

print(volgende(volgende(2)))
print(volgende_bis(volgende_bis(2)))
````

**Uitvoer:**
```
4
klaar
klaar
4
```

**Uitleg:**

**`return` beëindigt de functie onmiddellijk.** Alles wat na `return` staat (in dezelfde functie), wordt nooit uitgevoerd.

- In `volgende` staat `print("klaar")` **na** `return` → die regel wordt nooit bereikt.
- In `volgende_bis` staat `print("klaar")` **vóór** `return` → die regel wordt bij **elke** oproep uitgevoerd.

Bij een geneste oproep zoals `volgende(volgende(2))` werk je **van binnen naar buiten**: eerst de binnenste oproep, en het resultaat daarvan is het argument voor de buitenste.

**Regel 1: `print(volgende(volgende(2)))`**

| stap | oproep | wat gebeurt er? | uitvoer | teruggegeven |
|---|---|---|---|---|
| 1 | `volgende(2)` | `return 3` → functie stopt | – | 3 |
| 2 | `volgende(3)` | `return 4` → functie stopt | – | 4 |
| 3 | `print(4)` | | `4` | |

**Regel 2: `print(volgende_bis(volgende_bis(2)))`**

| stap | oproep | wat gebeurt er? | uitvoer | teruggegeven |
|---|---|---|---|---|
| 1 | `volgende_bis(2)` | eerst `print("klaar")`, dan `return 3` | `klaar` | 3 |
| 2 | `volgende_bis(3)` | eerst `print("klaar")`, dan `return 4` | `klaar` | 4 |
| 3 | `print(4)` | | `4` | |

- De twee keer `klaar` verschijnen **vóór** de `4`: `print(...)` kan pas iets tonen als de waarde tussen de haakjes volledig berekend is, en tijdens die berekening worden de twee `klaar`'s al getoond.

### Oefening 5

Voorspel de uitvoer. Verwacht je een foutmelding? Schrijf dan gewoon "fout".

````python
def verhoog(getal):
    """Getal verhogen met 10."""
    getal = getal + 10
    return getal

getal = 5
nieuw = verhoog(getal)
print(getal, nieuw)
````

**Uitvoer:**
```
5 15
```

**Uitleg:**

De parameter `getal` in de functie is een **lokale variabele**: een aparte variabele die enkel binnen de functie bestaat, ook al heeft ze dezelfde naam als de variabele `getal` buiten de functie.

| stap | `getal` (buiten) | `getal` (binnen de functie) | `nieuw` |
|---|---|---|---|
| `getal = 5` | 5 | – | – |
| oproep `verhoog(getal)` | 5 | 5 (kopie van de waarde) | – |
| `getal = getal + 10` (binnen) | 5 | 15 | – |
| `return getal` → `nieuw = 15` | 5 | (bestaat niet meer) | 15 |

- De variabele `getal` buiten de functie verandert **niet**: `print(getal, nieuw)` → `5 15`.


**Uitleg:**

| oproep | berekening | resultaat |

- `math.sqrt()` geeft **altijd een float**, ook als de wortel "opgaat": `5.0` en niet `5`.
- `int()` **kapt af**, het rondt niet af.

### Oefening 15 (bonusvraag – uitbreiding)

Voorspel de uitvoer. Verwacht je een foutmelding? Schrijf dan gewoon "fout".

````python
#uitbreiding (bonusvraag)
class Rechthoek:
    """Klasse Rechthoek."""

    def __init__(self, lengte, breedte):
        """Initialisatie: een rechthoek heeft een lengte en een breedte."""
        self.l = lengte
        self.b = breedte

    def oppervlakte(self):
        """Oppervlakte van de rechthoek."""
        return self.l * self.b

r1 = Rechthoek(4, 3)
r2 = Rechthoek(2, 5)
print(r1.oppervlakte() + r2.oppervlakte(), r1.l)
````

**Uitvoer:**
```
22 4
```

**Uitleg:**

- `Rechthoek(4, 3)` maakt een **object** aan: `__init__` wordt uitgevoerd met `lengte = 4` en `breedte = 3`, en die worden bewaard als attributen `self.l` en `self.b`.
- `self` verwijst naar het object zelf. Zo weet de methode `oppervlakte()` over welke rechthoek het gaat.

| object | `self.l` | `self.b` | `oppervlakte()` |
|---|---|---|---|
| `r1` | 4 | 3 | 4 · 3 = 12 |
| `r2` | 2 | 5 | 2 · 5 = 10 |

- `r1.oppervlakte() + r2.oppervlakte()` = 12 + 10 = `22`
- `r1.l` = het attribuut `l` van `r1` → `4`
- Uitvoer: `22 4`

### Oefening 16

**Zoek de fout!** Deze code bevat precies één fout. Verbeter ze.

Gewenste uitvoer: `24`

````python
def omtrek_vierkant(z)
    """Omtrek van een vierkant met zijde z."""
    return 4 * z

print(omtrek_vierkant(6))
````

**Uitvoer van de originele code:**
```
fout
```

**Verbeterde code:**

````python
def omtrek_vierkant(z):
    """Omtrek van een vierkant met zijde z."""
    return 4 * z

print(omtrek_vierkant(6))
````

**Uitleg:**

- Zonder dubbelepunt geeft Python een **fout**, nog voor er iets uitgevoerd wordt.
- Na de verbetering: `omtrek_vierkant(6)` = 4 · 6 = `24`.

### Oefening 17

**Zoek de fout!** Deze code bevat precies één fout. Verbeter ze.

Gewenste uitvoer: `Het dubbel van 7 is 14.`

````python
def verdubbelen(x):
    """Verdubbelen."""
    dubbel = 2 * x

print(f"Het dubbel van 7 is {verdubbelen(7)}.")
````

**Uitvoer van de originele code:**
```
Het dubbel van 7 is None.
```

**Verbeterde code:**

````python
def verdubbelen(x):
    """Verdubbelen."""
    dubbel = 2 * x
    return dubbel

print(f"Het dubbel van 7 is {verdubbelen(7)}.")
````

**Uitleg:**

- De functie berekent `dubbel = 14`, maar geeft niets terug: er ontbreekt `return dubbel`.
- Zonder `return` geeft de functie `None` terug, en dat wordt in de f-string ingevuld: `Het dubbel van 7 is None.`
- Geen foutmelding, maar een fout resultaat.
- Oplossing: `return dubbel` toevoegen (ingesprongen, in de functie). Korter mag ook: `return 2 * x`.

### Oefening 18

**Zoek de fout!** Deze code bevat precies één fout. Verbeter ze.

Gewenste uitvoer: `De prijs na korting is 65.0 euro.`

````python
def prijs_na_korting(pr, per):
    """Prijs na een korting van per procent op een prijs pr."""
    nieuwe_prijs = pr * (1 - pr / 100)
    return nieuwe_prijs

print(f"De prijs na korting is {prijs_na_korting(100, 35)} euro.")
````

**Uitvoer van de originele code:**
```
De prijs na korting is 0.0 euro.
```

**Verbeterde code:**

````python
def prijs_na_korting(pr, per):
    """Prijs na een korting van per procent op een prijs pr."""
    nieuwe_prijs = pr * (1 - per / 100)
    return nieuwe_prijs

print(f"De prijs na korting is {prijs_na_korting(100, 35)} euro.")
````

**Uitleg:**

- In de formule staat `pr / 100` in plaats van `per / 100`. De parameter `per` wordt nergens gebruikt!
- Originele code: 100 · (1 − 100/100) = 100 · 0 = `0.0` → `De prijs na korting is 0.0 euro.`
- Na de verbetering: 100 · (1 − 35/100) = 100 · 0,65 = `65.0`
- Tip: een parameter die nergens in de functie gebruikt wordt, is bijna altijd een teken van een fout.
