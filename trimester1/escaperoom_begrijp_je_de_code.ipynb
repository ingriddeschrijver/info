# Escape room: begrijp je de code?

Hieronder staat een klein voorbeeld van een escape room in Python, met twee kamers.

**Basisregels** die in deze code gevolgd worden:
- een functie krijgt alles wat ze nodig heeft via **parameters**;
- variabelen die in een functie ontstaan, zijn **lokaal**: ze bestaan enkel in die functie;
- een functie geeft haar resultaat terug met **`return`**;
- de lijst `inventaris` wordt enkel in het hoofdprogramma gebruikt en aangepast.

Speel het spel eerst een paar keer: voer de code-cel uit en probeer verschillende keuzes.

Beantwoord daarna de vragen. Probeer eerst zonder de code uit te voeren en controleer pas daarna. Vragen met ★ zijn moeilijker.

```python
# ESCAPE ROOM: Opgesloten in de school
# Voorbeeld met 2 kamers. Een groepje maakt er zelf 4 à 5.
#
# Basisregels:
# - een functie krijgt alles wat ze nodig heeft via parameters;
# - variabelen die in een functie ontstaan, zijn lokaal: ze bestaan enkel in die functie;
# - een functie geeft haar resultaat terug met return (een getal, een tekst, True/False);
# - de lijst inventaris wordt enkel in het hoofdprogramma gebruikt en aangepast.


def kamer_klaslokaal():
    """Eerste kamer: geeft het gevonden voorwerp terug, of een lege tekst als je niets vindt."""
    print("\n--- HET KLASLOKAAL ---")
    print("Je zit opgesloten in lokaal B12. Je ziet een bureau en een kast.")
    keuze = input("Wat onderzoek je? (bureau/kast) ").lower()
    if keuze == "bureau":
        print("In de lade ligt een sleutel! Je steekt hem in je zak.")
        return "sleutel"
    elif keuze == "kast":
        print("Enkel oude cursussen wiskunde... Je vindt niets.")
    else:
        print("Dat ken ik niet. Je verliest tijd.")
    return ""


def kamer_gang(heeft_sleutel):
    """Tweede kamer: deur met een cijferslot. Geeft True terug als de speler ontsnapt."""
    print("\n--- DE GANG ---")
    if not heeft_sleutel:
        print("De deur van het lokaal zit op slot. Je hebt een sleutel nodig!")
        return False

    print("Je opent de deur. Aan de uitgang hangt een cijferslot met een briefje:")
    print("'De code is de schuine zijde van een rechthoekige driehoek met zijden 20 en 21.'")
    pogingen = 0
    while pogingen < 3:
        poging = int(input("Code? "))
        pogingen += 1
        if poging == schuine_zijde(20, 21):
            print("KLIK! De deur gaat open.")
            return True
        print(f"Fout! Nog {3 - pogingen} poging(en).")
    print("Het alarm gaat af...")
    return False


def schuine_zijde(a, b):
    """Schuine zijde van een rechthoekige driehoek met rechthoekszijden a en b."""
    return int((a ** 2 + b ** 2) ** 0.5)


# ----- HOOFDPROGRAMMA -----
naam = input("Hoe heet je? ")
inventaris = []

voorwerp = kamer_klaslokaal()
if voorwerp != "":
    inventaris.append(voorwerp)

ontsnapt = kamer_gang("sleutel" in inventaris)

if ontsnapt:
    print(f"\nProficiat {naam}, je bent ontsnapt met: {inventaris}")
else:
    print(f"\nHelaas {naam}, je zit nog altijd vast. GAME OVER")
```
