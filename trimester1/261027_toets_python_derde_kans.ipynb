### Vraag 17 (4 pt)

Wat is de uitvoer?

```python
waarden = [6, 7, 2, 9, 11]
lijst = []
for w in waarden:
    if w % 3 == 0:
        lijst.append(w // 3)
        print("deelbaar:", w)
    elif w > 4:
        lijst.append(w + 1)
    else:
        print(f"te klein: {w}")
print(lijst, len(lijst))
```

### Vraag 21 (+3 pt bonus)

```python
for i in range(1, 4):
    lijn = ""
    for j in range(1, 5):
        if j <= i:
            lijn += str(j * i)
        else:
            lijn += "."
    print(lijn)
```
