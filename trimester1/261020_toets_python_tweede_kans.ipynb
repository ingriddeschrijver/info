### Vraag 17 (4 pt)

Wat is de uitvoer?

```python
getallen = [4, 7, 10, 3, 8]
resultaat = []
for g in getallen:
    if g % 2 == 0:
        resultaat.append(g // 2)
    elif g > 5:
        resultaat.append(g * 10)
    else:
        print(f"overslaan: {g}")
    print(len(resultaat))
print(resultaat)
```
