### Oefening 11

````python
getallen = [7, 3, 12, 5, 13, 11, 8]
grootste = getallen[0]
print(f'Tussenstand: grootste = {grootste}')
for g in getallen:
    if g > grootste:
        grootste = g
        print(f'Tussenstand: grootste = {grootste}')
print(f'Het grootste getal van de lijst is {grootste}.')
````

### Oefening 12
