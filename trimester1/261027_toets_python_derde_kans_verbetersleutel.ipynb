## Deel 1: Datatypes (12 pt)

### Vraag 1 (2 pt)

Wat is de uitvoer?

````python
x = 7.0
y = 2
print(x // y, x % y)
print(type(x // y), int(x) / y)
````

**Uitvoer:**
```
3.0 1.0
<class 'float'> 3.5
```

**Uitleg:**

- `x = 7.0` is een **float** (er staat een punt), `y = 2` is een **int**.
- Zodra er een float meedoet in een bewerking, is het resultaat ook een **float**, ook bij `//` en `%`.

| uitdrukking | berekening | resultaat | type |
|---|---|---|---|
| `x // y` | gehele deling: 2 past 3 keer in 7 | `3.0` | float |
| `x % y` | rest: 7 = 3 · 2 + **1** | `1.0` | float |
| `type(x // y)` | het type van `3.0` | `<class 'float'>` | – |
| `int(x)` | `7.0` → `7` | `7` | int |
| `int(x) / y` | 7 / 2 (gewone deling) | `3.5` | float |

- ⚠️ Valkuil: `x // y` geeft **niet** `3` maar `3.0`, omdat `x` een float is.
- `/` geeft altijd een float, ook met twee integers.

- `str(b) * 4` = `"24"` 4 keer herhalen → `"24242424"` → `len(...)` telt de tekens: **8**

| index | 0 | 1 | 2 | 3 |

### Vraag 10 (2 pt)

Wat is de uitvoer?

````python
a = 20 - 3 * 4 // 5 + 2 ** 3
b = (20 - 3) * 4 // (5 + 2) ** 2
c = 7 % 3 * 2 + 10 // 4
print(a, b, c)
````

**Uitvoer:**
```
26 1 4
```

**Uitleg:**

Volgorde van bewerkingen: **haakjes → `**` → `*`, `/`, `//`, `%` → `+`, `-`** (bij gelijke voorrang van links naar rechts).

| variabele | berekening stap voor stap | resultaat |
|---|---|---|
| `a = 20 - 3 * 4 // 5 + 2 ** 3` | `2 ** 3` = 8 → `3 * 4` = 12 → `12 // 5` = 2 → 20 − 2 + 8 | **26** |
| `b = (20 - 3) * 4 // (5 + 2) ** 2` | haakjes: 17 en 7 → `7 ** 2` = 49 → `17 * 4` = 68 → `68 // 49` = 1 | **1** |
| `c = 7 % 3 * 2 + 10 // 4` | `7 % 3` = 1 → `1 * 2` = 2 → `10 // 4` = 2 → 2 + 2 | **4** |

- `*`, `//` en `%` hebben **dezelfde voorrang**: dan reken je gewoon van links naar rechts (`3 * 4` eerst, daarna `// 5`).
- `**` gaat vóór `//`: in `b` wordt eerst `(5 + 2) ** 2` = 49 berekend, pas dan `68 // 49`.

### Vraag 11 (3 pt)

Wat is de uitvoer?

````python
zon = False
uur = 14
print(zon or uur >= 12 and uur < 18)
print(not (zon or uur % 7 == 0))
print(not zon and uur > 15 or uur == 14)
````

**Uitvoer:**
```
True
False
True
```

**Uitleg:**

`zon = False`, `uur = 14`

Volgorde bij logische operatoren: **vergelijkingen → `not` → `and` → `or`**.

| regel | stap 1 (vergelijkingen) | stap 2 | resultaat |
|---|---|---|---|
| `zon or uur >= 12 and uur < 18` | `14 >= 12` = True; `14 < 18` = True | `and` eerst: `True and True` = True → `False or True` | `True` |
| `not (zon or uur % 7 == 0)` | `14 % 7` = 0 → `0 == 0` = True | haakjes: `False or True` = True → `not True` | `False` |
| `not zon and uur > 15 or uur == 14` | `not False` = True; `14 > 15` = False; `14 == 14` = True | `and` eerst: `True and False` = False → `False or True` | `True` |

- `and` gaat vóór `or` (net zoals `*` vóór `+`).
- `or` is True zodra **één** kant True is; `and` is pas True als **beide** kanten True zijn.

## Deel 3: Structuren (13 pt)

### Vraag 15 (1 pt)

Wat is de uitvoer?

````python
n = 100
while n > 10:
    n = n // 3
print(n)
````

**Uitvoer:**
```
3
```

**Uitleg:**

Een `while`-lus controleert de voorwaarde **vóór elke doorloop**. Zolang ze True is, wordt het ingesprongen blok uitgevoerd.

| doorloop | `n` bij de controle | `n > 10`? | berekening | nieuwe `n` |
|---|---|---|---|---|
| 1 | 100 | True | 100 // 3 | 33 |
| 2 | 33 | True | 33 // 3 | 11 |
| 3 | 11 | True | 11 // 3 | 3 |
| – | 3 | **False** → stop | – | – |

- ⚠️ Valkuil: bij `n = 11` is `11 > 10` nog **True**, dus er volgt nog een doorloop. Wie stopt bij 11, heeft het fout.
- `print(n)` staat **niet ingesprongen** → wordt pas na de lus één keer uitgevoerd: `3`.

### Vraag 16 (2 pt)

Wat is de uitvoer?

````python
zin = "banaan"
punten = 0
for letter in zin:
    if letter == "a":
        punten += 2
    elif letter == "n":
        punten -= 1
    else:
        punten += 5
print(punten)
````

**Uitvoer:**
```
9
```

**Uitleg:**

De `for`-lus overloopt de tekst **letter per letter**. Per letter wordt precies **één** van de drie takken uitgevoerd.

| doorloop | `letter` | welke tak? | actie | `punten` na deze doorloop |
|---|---|---|---|---|
| start | – | – | – | 0 |
| 1 | b | `else` | +5 | 5 |
| 2 | a | `if` | +2 | 7 |
| 3 | n | `elif` | −1 | 6 |
| 4 | a | `if` | +2 | 8 |
| 5 | a | `if` | +2 | 10 |
| 6 | n | `elif` | −1 | 9 |

- Controle: 1 × `b` (+5), 3 × `a` (+6), 2 × `n` (−2) → 5 + 6 − 2 = **9**
- `print(punten)` staat na de lus → er wordt maar één getal getoond.

### Vraag 17 (4 pt)

Wat is de uitvoer?

````python
waarden = [6, 7, 2, 9, 11]
lijst = []
for w in waarden:
    if w % 3 == 0:
        lijst.append(w // 3)
        print("deelbaar:", w)
    elif w > 4:
        lijst.append(w + 1)
    else:
        print(f"te klein: {w}")
print(lijst, len(lijst))
````

**Uitvoer:**
```
deelbaar: 6
te klein: 2
deelbaar: 9
[2, 8, 3, 12] 4
```

**Uitleg:**

Let goed op de inspringing:
- `print("deelbaar:", w)` staat **in de `if`** → enkel bij getallen deelbaar door 3.
- In de `elif` wordt er **niets** getoond, enkel iets toegevoegd.
- `print(lijst, len(lijst))` staat **na de lus** → één keer, helemaal op het einde.

| doorloop | `w` | `w % 3` | `w % 3 == 0`? | `w > 4`? | welke tak? | toegevoegd | `lijst` na deze doorloop | uitvoer |
|---|---|---|---|---|---|---|---|---|
| start | – | – | – | – | – | – | `[]` | – |
| 1 | 6 | 0 | True | (niet gecontroleerd) | `if` | 6 // 3 = 2 | `[2]` | `deelbaar: 6` |
| 2 | 7 | 1 | False | True | `elif` | 7 + 1 = 8 | `[2, 8]` | – |
| 3 | 2 | 2 | False | False | `else` | – | `[2, 8]` | `te klein: 2` |
| 4 | 9 | 0 | True | (niet gecontroleerd) | `if` | 9 // 3 = 3 | `[2, 8, 3]` | `deelbaar: 9` |
| 5 | 11 | 2 | False | True | `elif` | 11 + 1 = 12 | `[2, 8, 3, 12]` | – |

- Na de lus: `print(lijst, len(lijst))` → `[2, 8, 3, 12] 4`
- `print("deelbaar:", w)` gebruikt een komma → Python zet zelf een spatie na de dubbele punt: `deelbaar: 6`.
- Bij 6 en 9 is ook `w > 4` waar, maar dat wordt niet meer gecontroleerd: de `if` was al True.

### Vraag 18 (2 pt)

Wat is de uitvoer? Geef de uitvoer voor elk van deze invoeren: "15" en "12".

````python
punt = int(input("Punt? "))
if punt >= 10:
    print("Geslaagd")
if punt >= 14:
    print("Onderscheiding")
elif punt >= 8:
    print("Bijna")
else:
    print("Niet geslaagd")
````

**Uitvoer bij invoer "15":**
```
Geslaagd
Onderscheiding
```

**Uitvoer bij invoer "12":**
```
Geslaagd
Bijna
```

**Uitleg:**

⚠️ De valkuil: er staan **twee aparte keuzestructuren** na elkaar.

1. Een losse `if punt >= 10` (zonder elif/else).
2. Een nieuwe `if punt >= 14 / elif punt >= 8 / else`.

De tweede `if` is **geen** `elif` van de eerste! Python controleert ze dus **allebei**, los van elkaar. Er kunnen dus twee regels uitvoer verschijnen.

| controle | `punt = 15` | `punt = 12` |
|---|---|---|
| **structuur 1:** `punt >= 10` | True → `Geslaagd` | True → `Geslaagd` |
| **structuur 2:** `punt >= 14` | **True** → `Onderscheiding`, rest van structuur 2 overslaan | False |
| `elif punt >= 8` | (niet meer gecontroleerd) | **True** → `Bijna`, stop |
| `else` | – | – |

- Bij 12 is `Bijna` eigenlijk een vreemde boodschap na `Geslaagd`: het programma zit logisch slecht in elkaar, maar we voorspellen wat Python **doet**, niet wat de bedoeling was.
- Wie bij een invoer maar één regel schrijft, heeft de tweede `if` als een `elif` gelezen.

### Vraag 19 (2 pt)

- Door `end=" "` komt alles op **één regel**, met een spatie in plaats van een nieuwe regel. Daarna volgt `klaar` op dezelfde regel: `2 4 6 8 10 klaar`.

### Vraag 20 (2 pt)

**Zoek de fout!** Deze code bevat precies één fout. Verbeter ze.

````python
woorden = ["appel", "ei", "banaan", "kiwi"]
for w in woorden:
    lang = 0
    if len(w) > 3:
        lang += 1
print(f"Lange woorden: {lang}")
````

**Gewenste uitvoer:** `Lange woorden: 3`

**Uitvoer van de originele code:**
```
Lange woorden: 1
```

**Verbeterde code:**

````python
woorden = ["appel", "ei", "banaan", "kiwi"]
lang = 0
for w in woorden:
    if len(w) > 3:
        lang += 1
print(f"Lange woorden: {lang}")
````

**Uitleg:**

- `lang = 0` staat **in de lus**: bij **elk** woord wordt de teller opnieuw op 0 gezet. Daardoor onthoudt `lang` enkel of het **laatste** woord lang is.
- Er komt geen foutmelding, maar een fout resultaat: `Lange woorden: 1`.

Verloop van de **originele** code:

| doorloop | `w` | `lang = 0` | `len(w)` | `len(w) > 3`? | `lang` na deze doorloop |
|---|---|---|---|---|---|
| 1 | appel | 0 | 5 | True | 1 |
| 2 | ei | **0** (gewist!) | 2 | False | 0 |
| 3 | banaan | **0** (gewist!) | 6 | True | 1 |
| 4 | kiwi | **0** (gewist!) | 4 | True | 1 |

- Oplossing: `lang = 0` **vóór de lus** zetten (niet ingesprongen). Een teller begin je altijd één keer, vóór de herhaling.

Verloop **na de verbetering**:

| doorloop | `w` | `len(w)` | `len(w) > 3`? | `lang` na deze doorloop |
|---|---|---|---|---|
| start | – | – | – | 0 |
| 1 | appel | 5 | True | 1 |
| 2 | ei | 2 | False | 1 |
| 3 | banaan | 6 | True | 2 |
| 4 | kiwi | 4 | True | 3 |

→ `Lange woorden: 3`

## Bonusvraag (geneste herhaling) (+3 pt bonus)

### Vraag 21 (+3 pt bonus)

Wat is de uitvoer?

````python
for i in range(1, 4):
    lijn = ""
    for j in range(1, 5):
        if j <= i:
            lijn += str(j * i)
        else:
            lijn += "."
    print(lijn)
````

**Uitvoer:**
```
1...
24..
369.
```

**Uitleg:**

- De buitenste lus loopt over `i` = 1, 2, 3 (4 niet).
- Per `i` begint `lijn` opnieuw **leeg** en loopt de binnenste lus over `j` = 1, 2, 3, 4.
- Is `j <= i`, dan wordt het **product** `j * i` als tekst toegevoegd; anders een punt.
- `print(lijn)` staat in de buitenste lus (niet in de binnenste) → één regel per `i`.

**`i = 1`**

| `j` | `j <= 1`? | toegevoegd | `lijn` |
|---|---|---|---|
| 1 | True | `str(1 * 1)` = `"1"` | `1` |
| 2 | False | `"."` | `1.` |
| 3 | False | `"."` | `1..` |
| 4 | False | `"."` | `1...` |

**`i = 2`**

| `j` | `j <= 2`? | toegevoegd | `lijn` |
|---|---|---|---|
| 1 | True | `str(1 * 2)` = `"2"` | `2` |
| 2 | True | `str(2 * 2)` = `"4"` | `24` |
| 3 | False | `"."` | `24.` |
| 4 | False | `"."` | `24..` |

**`i = 3`**

| `j` | `j <= 3`? | toegevoegd | `lijn` |
|---|---|---|---|
| 1 | True | `str(1 * 3)` = `"3"` | `3` |
| 2 | True | `str(2 * 3)` = `"6"` | `36` |
| 3 | True | `str(3 * 3)` = `"9"` | `369` |
| 4 | False | `"."` | `369.` |

- Omdat `lijn` tekst is, worden de cijfers **aan elkaar geplakt** (`"2" + "4"` = `"24"`), niet opgeteld.
- Elke regel heeft precies 4 tekens, want de binnenste lus draait altijd 4 keer.
