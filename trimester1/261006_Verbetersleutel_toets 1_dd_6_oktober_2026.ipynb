## 🏷️ Overzicht van de gebruikte afkortingen in de feedback
Naast een fout antwoord kan je een van onderstaande afkortingen terugvinden. Zo weet je precies wat er misliep.  
Bij sommige afkortingen staat een voorbeeld dat je zelf kan uitvoeren.

### PK – *print-komma*
Bij de uitvoer wordt er een komma geschreven die er in werkelijkheid niet komt te staan. De komma’s in het print-commando dienen enkel om de verschillende argumenten van elkaar te scheiden; in de uitvoer komt er een **spatie** tussen.

````python
a = 3.5
b = 'woord'
print(a, b, 'Joepie!')     # → 3.5 woord Joepie!
````

### PDT – *print-decimaal teken*
In Python wordt een **punt** (`.`) gebruikt als decimaal teken en geen komma.

### PH – *print-haakjes*
De haakjes die je in de code rondom de argumenten van het print-commando plaatst (zoals de haakjes rondom `a, b, 'Joepie!'`), worden vanzelfsprekend **niet** mee geprint.

### PNL – *print-nieuwe lijn*
De argumenten binnen één enkel print-commando worden netjes op **één lijn** geprint. Plaats je meerdere print-commando’s onder elkaar in de code, dan start elk commando op een **nieuwe lijn** (tenzij je expliciet `end=''` toevoegt).

### PNV – *print-naam variabele*
Je hebt de **naam** van de variabele zelf genoteerd als uitvoer in plaats van de **inhoud** (waarde) die in die variabele is opgeslagen. Dit werd wel foutief aangerekend.

### PAT – *print-aanhalingstekens*
Je noteert aanhalingstekens in de uitvoer waar ze niet thuishoren, of omgekeerd. Ter verduidelijking:

````python
a = "12"     # in a wordt een string (tekst) opgeslagen
print(a)     # → 12  (Er worden geen aanhalingstekens geprint!)
print(a*3)   # → 121212
````

### LS – *list-syntax*
Wanneer je een lijst print, wordt deze omsloten door **rechte haakjes** `[ ]` en worden de elementen van de lijst gescheiden door een **komma**.

### SS – *string-syntax*
Je plaatst geen aanhalingstekens rondom een string wanneer deze deel uitmaakt van een lijst. Let op het verschil:

````python
print('a')        # → a   (zonder aanhalingstekens)
x = 'een'
y = 'twee'
lijst = [x, y]
print(lijst)      # → ['een', 'twee']
````

### PF – *print-format / f-strings*
Je maakte een fout bij het letterlijk overnemen of invullen van de tekst bij een print-commando met formattering (f-strings). Alles wat **niet** tussen accolades `{ }` staat, wordt letterlijk overgenomen in de uitvoer.

````python
a = 'een'
b = 'twee'
c = 3.5
print(f'a,#{b}@[{c}]')     # → a,#twee@[3.5]
````

### Vraag 10  <small>(2 pt)</small>

````python
print(17 // 3 * 3 + 17 % 3)
print(-7 // 2, -7 % 2)
````

**Uitvoer:**
```

```

#### 🔎 Stap voor stap

````python
q = 17 // 3           # gehele deling → 5
r = 17 % 3            # rest → 2   (want 17 = 3 * 5 + 2)
p = q * 3             # // en * hebben dezelfde voorrang: van links naar rechts → 15
print(p + r)          # 15 + 2 → 17
q2 = -7 // 2          # -3.5 afronden naar BENEDEN → -4
r2 = -7 % 2           # want -7 = 2 * (-4) + 1 → 1
print(q2, r2)         # → -4 1
````

#### ➕ Extra toelichting bij de restverschijnselen ( // en % )
De tweede regel van de code bij vraag 10 was uitbreiding en stond niet op punten.  
Ter info: bij modulo-rekenen in Python heeft de rest steeds hetzelfde teken als de deler. (In andere programmeertalen is dit niet altijd het geval.)

| // (quotiënt) | % (rest) | Want … |
|:---:|:---:|---|
| `7 // 2` → **3** | `7 % 2` → **1** | `7 = 2 * 3 + 1`<br>*de rest heeft hetzelfde teken als de deler* |
| `-7 // 2` → **-4** | `-7 % 2` → **1** | `-7 = 2 * (-4) + 1`<br>*de rest moet positief zijn, gelijk aan het teken van de deler* |
| `7 // -2` → **-4** | `7 % -2` → **-1** | `7 = -2 * (-4) + (-1)` |
| `-7 // -2` → **3** | `-7 % -2` → **-1** | `-7 = -2 * 3 + (-1)` |

````python
for a, b in [(7, 2), (-7, 2), (7, -2), (-7, -2)]:
    q = a // b
    r = a % b
    print(f"{a} // {b} = {q}    {a} % {b} = {r}    controle: {b} * {q} + {r} = {b * q + r}")
````

> **Algemeen:**
> - `a == b * (a // b) + (a % b)` geldt altijd.
> - De rest `a % b` heeft in Python steeds hetzelfde teken als de deler `b`.
> - `a // b` is gelijk aan het grootste geheel getal dat kleiner dan of gelijk aan `a / b` is (bv. `7 / 2 = 3.5` → `7 // 2 = 3`, `-7 / 2 = -3.5` → `-7 // 2 = -4`, maar `6 / 2 = 3.0` → `6 // 2 = 3`).

### Vraag 17  <small>(4 pt)</small>
Wat is de uitvoer?

````python
n = 2
stappen = 0
while n != 1:
    if n % 2 == 0:
        n = n // 2
        print(f'true {n}')
    else:
        n = 3 * n + 1
        print(f'false {n}')

    stappen += 1
    print(stappen)
print(f"{stappen} stappen")
````

**Uitvoer:**
```
true 1
1
1 stappen
```

#### 🔎 Stap voor stap

| ronde | n != 1 ? | n % 2 == 0 ? | n wordt | stappen | uitvoer |
|:---:|:---:|:---:|:---:|:---:|:---:|
| 1 | 2 != 1 → True | 2 % 2 == 0 → True | 2 // 2 = 1 | 1 | true 1<br>1 |
| 2 | 1 != 1 → False | — | — | — | lus stopt |


De `else`-tak wordt nooit uitgevoerd, dus `false …` verschijnt niet. Na de lus volgt nog `print(f"{stappen} stappen")` → **`1 stappen`**.

**Alternatief: dezelfde code, maar start met** `n = 3`**.** Nu worden beide takken gebruikt: is n even, dan wordt n gehalveerd (`true`); is n oneven, dan wordt n = 3 · n + 1 (`false`).

````python
n = 3   # probeer ook eens 6 of 7!
stappen = 0
while n != 1:
    if n % 2 == 0:
        n = n // 2
        print(f'true {n}')
    else:
        n = 3 * n + 1
        print(f'false {n}')

    stappen += 1
    print(stappen)
print(f"{stappen} stappen")
````

| ronde | n != 1 ? | n % 2 == 0 ? | n wordt | stappen | uitvoer |
|:---:|:---:|:---:|:---:|:---:|:---:|
| 1 | 3 != 1 → True | 3 % 2 == 0 → False | 3 * 3 + 1 = 10 | 1 | false 10<br>1 |
| 2 | 10 != 1 → True | 10 % 2 == 0 → True | 10 // 2 = 5 | 2 | true 5<br>2 |
| 3 | 5 != 1 → True | 5 % 2 == 0 → False | 3 * 5 + 1 = 16 | 3 | false 16<br>3 |
| 4 | 16 != 1 → True | 16 % 2 == 0 → True | 16 // 2 = 8 | 4 | true 8<br>4 |
| 5 | 8 != 1 → True | 8 % 2 == 0 → True | 8 // 2 = 4 | 5 | true 4<br>5 |
| 6 | 4 != 1 → True | 4 % 2 == 0 → True | 4 // 2 = 2 | 6 | true 2<br>6 |
| 7 | 2 != 1 → True | 2 % 2 == 0 → True | 2 // 2 = 1 | 7 | true 1<br>7 |
| 8 | 1 != 1 → False | — | — | — | lus stopt |


Na de lus: `print(f"{stappen} stappen")` → **`7 stappen`**. De rij waarden van n is dus 3 → 10 → 5 → 16 → 8 → 4 → 2 → 1.

### Vraag 18  <small>(2 pt)</small>
Wat is de uitvoer? Geef de uitvoer voor elk van deze invoeren: "15" en "9".

> ⌨️ Typ bij het uitvoeren: "15" (en daarna nog eens met "9")

````python
punten = int(input("Punten op 20? "))
if punten >= 16:
    oordeel = "zeer goed"
elif punten >= 10:
    oordeel = "voldoende"
elif punten >= 14:
    oordeel = "goed"
else:
    oordeel = "onvoldoende"
print(f"{punten}/20: {oordeel} ({punten * 5}%)")
````

**Uitvoer:**
```
Bij invoer "15":  15/20: voldoende (75%)
Bij invoer "9":   9/20: onvoldoende (45%)
```

#### 🔎 Stap voor stap

| invoer | punten >= 16 | punten >= 10 | punten >= 14 | oordeel | punten * 5 |
|:---:|:---:|:---:|:---:|:---:|:---:|
| 15 | False | True  ✔ | niet getest | voldoende | 75 |
| 9 | False | False | False | onvoldoende | 45 |


Bij `if … elif … else` wordt **enkel de eerste ware voorwaarde** uitgevoerd; daarna slaat Python de rest over. Omdat `punten >= 10` vóór `punten >= 14` staat, kan "goed" nooit voorkomen.

````python
tekst = input("Punten op 20? ")   # tekst = "15"
punten = int(tekst)               # → 15
print(punten >= 16)               # → False
print(punten >= 10)               # → True  → oordeel = "voldoende"
procent = punten * 5              # → 75
print(f"{punten}/20: voldoende ({procent}%)")   # → 15/20: voldoende (75%)
````

### Vraag 19  <small>(2 pt)</small>
Zoek de fout! Deze code bevat precies één fout. Verbeter ze. De gebruiker typt: "robot".  
**Gewenste uitvoer:** `Omgekeerd: tobor`

> ⌨️ Typ bij het uitvoeren: "robot"

````python
woord = input("Woord? ")
omgekeerd = ""
for letter in woord:
    omgekeerd = omgekeerd + letter
print(f"Omgekeerd: {omgekeerd}")
````

**Verbeterde regel(s):**
```
omgekeerd = letter + omgekeerd
```
*De letter moet vóór de tekst komen, anders blijft het woord gewoon robot.*

#### 🔎 Stap voor stap

| letter | fout: omgekeerd + letter | juist: letter + omgekeerd |
|:---:|:---:|:---:|
| r | r | r |
| o | ro | or |
| b | rob | bor |
| o | robo | obor |
| t | robot | tobor |


Tip: zet in de lus een extra `print(letter, omgekeerd)` om elke tussenstap te zien.

**De verbeterde code, met een extra print in de lus:**

````python
woord = input("Woord? ")
omgekeerd = ""
for letter in woord:
    omgekeerd = letter + omgekeerd
    print(letter, omgekeerd)      # extra print: elke tussenstap
print(f"Omgekeerd: {omgekeerd}")
````

### Vraag 21  <small>(3 pt)</small>
Wat is de uitvoer?

````python
for i in range(1, 4):
    rij = ""
    for j in range(i, 4):
        rij += str(i * j) + " "
    print(f"{i}: {rij}")
````

**Uitvoer:**
```
1: 1 2 3
2: 4 6
3: 9
```

#### 🔎 Stap voor stap

````python
for i in range(1, 4):           # i = 1, 2, 3
    rij = ""                    # bij elke nieuwe i: lege rij
    for j in range(i, 4):       # j start bij i, stopt vóór 4
        rij += str(i * j) + " " # getal → tekst + spatie toevoegen
        print(i, j, rij)        # extra print: toont elke tussenstap
    print(f"{i}: {rij}")
````

| i | range(i, 4) | j | i * j | rij na deze stap | uitvoer |
|:---:|:---:|:---:|:---:|:---:|:---:|
| 1 | 1, 2, 3 | 1 | 1 | "1 " |   |
|   |   | 2 | 2 | "1 2 " |   |
|   |   | 3 | 3 | "1 2 3 " | 1: 1 2 3 |
| 2 | 2, 3 | 2 | 4 | "4 " |   |
|   |   | 3 | 6 | "4 6 " | 2: 4 6 |
| 3 | 3 | 3 | 9 | "9 " | 3: 9 |
