### Vraag 17 (4 pt)

Wat is de uitvoer?

````python
n = 2
stappen = 0
while n != 1:
    if n % 2 == 0:
        n = n // 2
        print(f'true {n}')
    else:
        n = 3 * n + 1
        print(f'false {n}')

    stappen += 1
    print(stappen)
print(f"{stappen} stappen")
````
