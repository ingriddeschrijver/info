```python
prijs_zonder_btw = float(input('Wat is het kostprijs zonder btw? '))
btw_percentage = int(input('Wat is het btw-percentage? '))
btw_factor = btw_percentage/100+1
prijs_inclusief_btw = prijs_zonder_btw * btw_factor
print(f'prijs_exclusief = {prijs_zonder_btw} | btw-percentage = {btw_percentage}  |  prijs_inclusief = {prijs_inclusief_btw} ')
```
