```python
#verdieping
voornamen = ["De", "Van", "Ter"]
achternamen = ["Boer", "Dijk", "Berg"]
combinaties = []
for deel1, deel2 in zip(voornamen, achternamen):
    combinaties.append(deel1 + " " + deel2)
print(combinaties)
```

```python
woord = "programmeren"
klinkers = "aeiou"
aantal = 0
for letter in woord:
    if letter in klinkers:
        aantal += 1
        print(letter , end = '+')
print()
print(aantal)
```

```python
tekst = "bananen"
klinkers = "aeiou"
resultaat = ""
for letter in tekst:
    if letter in klinkers:
        resultaat = resultaat + letter.upper()
    else:
        resultaat = resultaat + letter
print(resultaat)
```
