# Patronen printen – voorspel de uitvoer
Informatica-wetenschappen · I. De Schrijver - gemaakt met behulp van AI Claude

**Opdracht:** voorspel bij elke oefening de uitvoer **zonder de code uit te voeren**, net zoals op de toets. Noteer je antwoord in het vak *Jouw voorspelling* (dubbelklik om te typen) of op papier. Pas daarna voer je de codecel uit (Shift + Enter) om te controleren.

De oefeningen worden steeds moeilijker: van ⭐ tot ⭐⭐⭐⭐⭐, met op het einde een paar bonusoefeningen.

**Spiekbriefje**
| code | betekenis |
|---|---|
| `range(1, 6)` | 1, 2, 3, 4, 5 (de 6 doet niet mee) |
| `range(3)` | 0, 1, 2 (start bij 0!) |
| `range(1, 8, 2)` | 1, 3, 5, 7 (stappen van 2) |
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 |
| `"ab" * 3` | `ababab` |
| `print(x, end="")` | print x en blijf op dezelfde lijn |
| `print()` | ga naar een nieuwe lijn |
| `i % 2` | rest bij deling door 2: 0 (even) of 1 (oneven) |
| `j % 3` | rest bij deling door 3: 0, 1 of 2 |

💡 **Tip:** maak bij een lus in een lus een tabelletje met de waarden van `i` en `j`. Noteer een spatie op papier als `·`, dan vergeet je ze niet.

*De patronen zijn gebaseerd op de "number pattern programs" van codeforwin.org (daar in C), herschreven in Python.*

## Niveau 4 – voorwaarden in de lus

### Oefening 18  ⭐⭐⭐⭐
Wat is de uitvoer?

````python
rijen = 4
kolommen = 7
for i in range(1, rijen + 1):
    for j in range(1, kolommen + 1):
        if i == 1 or i == rijen or j == 1 or j == kolommen:
            print(1, end="")
        else:
            print(0, end="")
    print()
````
